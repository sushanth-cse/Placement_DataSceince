```python
import numpy as np

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],
    [2, 120000, 140000, 160000, 190000],
    [3, 200000, 230000, 260000, 300000],
    [4, 180000, 210000, 240000, 270000],
    [5, 160000, 185000, 205000, 230000]
])
sales = sales_data[:, 1:]
print("Shape:", sales_data.shape)
print("\nFirst 3 restaurants:")
print(sales_data[:3])
yearly_total = sales.sum(axis=0)
print("\nTotal sales every year:", yearly_total)
minimum_sales = sales.min(axis=1)
print("\nMinimum sale per restaurant:", minimum_sales)
maximum_sales = sales.max(axis=0)
print("\nMaximum sales per year:", maximum_sales)
average_per_restaurant = sales.mean(axis=1)
print("\nAverage sales per restaurant:", average_per_restaurant)
average_cumulative_sale = sales.mean()
print("\nAverage cumulative sale:", average_cumulative_sale)
monthly_average = yearly_total / 12
print("\nMonthly average every year:", monthly_average)
```

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])
# Addition
print("Addition:", a + b)
# Subtraction
print("Subtraction:", a - b)
# Multiplication
print("Multiplication:", a * b)
# Division
print("Division:", a / b)
# Power
print("Power:", a ** 2)
# Square root
print("Square Root:", np.sqrt(a))
```
