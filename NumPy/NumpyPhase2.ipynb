```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

# Addition
print("Addition:", a + b)

# Subtraction
print("Subtraction:", a - b)

# Multiplication
print("Multiplication:", a * b)

# Division
print("Division:", a / b)

# Power
print("Power:", a ** 2)

# Square root
print("Square Root:", np.sqrt(a))
```
