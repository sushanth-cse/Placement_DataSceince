Filter

```python
nums=np.array([4,2,9,6,3,5,7,3])
even_nums=nums[nums%2==0]
print("Even Numbers",even_nums)
#Filter with mask
cgpa=np.array([4,3.2,7,8.9,9,8,5,4.3,7.9,9.0])
res=cgpa>4.5
print(res)
```

```python
#join 2d array (vertical and horizontal) 
matrix=np.array([10,20],
                [30,40])
newMatrix=np.array([[50,60],
                    [70,80]])
print("Horizontal:\n",np.hpstack([matrix, newMatrix]))
print("Vertical:\n",np.vpstack([matrix, newMatrix]))

```
