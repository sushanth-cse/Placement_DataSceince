Numpy Function

```python
import numpy as np
```

Aggregate Funtions

```python
arr=np.array([10,20,30,40,50])
print("Sum:",np.sum(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
print("Comulative Sum:",np.cumsum(arr))
```

Stastical function

```python
import statistics as st
m1=np.array([18,22,17,25,20,18,23])
m2=np.array([20,15,15,16,35,13,19])
print("Mean M1",np.mean(m1),"Means m2:",np.mean(m2))
print("Median M1",np.median(m1),"Means m2:",np.median(m2))
print("Mode M1",st.mode(m1),"Mode m2:",st.mode(m2))
print("Standard deviation",np.std(m1),"Standard deviation m2:",np.std(m2))
```
