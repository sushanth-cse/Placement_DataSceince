```python
import pandas as pd
```

##GroupBy

```python
exp=pd.read_excel("Dataset_Pandas/expense.xlsx")
exp                                     
```

```python
#Find the total expense for each category
exp.groupby('Category')['Amount'].sum() #M1
exp.groupby('Category')['Amount'].agg(['sum']) #M2
#Give a complete expense summary for each category.
exp.groupby('Category')['Amount'].agg(['sum','count','mean','max','min']) #M2
#3.  How much money was paid by each payment mode
exp.groupby('Payment Mode')['Amount'].agg(['sum']) #M2
#3.  How many transaction made using each payment mode
exp.groupby('Payment Mode')['Amount'].agg(['sum','count']) #M2

```

```python
#5.For  each payment mode,calculate the total spent across categories
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum'])
#5.For  each payment mode,calculate the total spent across Payment Mode
exp.groupby(['Category','Payment Mode'])['Amount'].agg(['sum'])

#Which categories have the  highest total expenses?
exp.groupby('Category')['Amount'].sum().sort_values
```

#MergeOrJoin

```python
customer=pd.read_csv('Dataset_Pandas/customer.csv')
customer
payment=pd.read_csv('Dataset_Pandas/payment.csv')
```

```python
#-------------Join------------
#Inner join
pd.merge(left=customer,right=payment,on="customerid",how="inner")
#Left join
pd.merge(left=customer,right=payment,on="customerid",how="left")
#Right
pd.merge(left=customer,right=payment,on="customerid",how="right")

```
