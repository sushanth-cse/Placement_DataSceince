Compare,Pivoting & Melting

```python
import pandas as pd
```

```python
shop_Oct8={
    "fruits":['mango','apple','banana','papaya'],
    'Price':[100,150,60,30],
    "quint":[10,8,12,15]
    }
stock_Oct8=pd.DataFrame(shop_Oct8)
shop_Oct9={
    "fruits":['mango','apple','banana','papaya'],
    'Price':[120,200,60,35],
    "quint":[15,6,24,13]
    }
stock_Oct9=pd.DataFrame(shop_Oct9)
#---------------Compare---------------#
stock_Oct8.compare(stock_Oct9)
stock_Oct8.compare(stock_Oct9,align_axis=0)
stock_Oct8.compare(stock_Oct9,align_axis=0,keep_equal=True)
stock_Oct8.compare(stock_Oct9,align_axis=0,keep_equal=True,keep_shape=True)
```

```python
monthly_sales={
    "Products":["laptop","mobile","laptop","mobile","laptop","laptop"],
    "Month":['Jan','Jan','Feb','Feb','Mar','Mar'],
    "Price":[65000,25000,78500,17500,75000,45000]
}
sales=pd.DataFrame(monthly_sales)
sales

#--------Pivoting--------#
sales.pivot(
    columns="Month",
    index="Products",
    values="Price"
    )

```

```python
monthly_sales={
    "Products":["laptop","mobile","laptop","mobile","laptop","laptop"],
    "Month":['Jan','Jan','Feb','Feb','Mar','Mar'],
    "Price":[65000,25000,78500,17500,75000,45000]
}
sales=pd.DataFrame(monthly_sales)
sales

#--------Pivoting--------#
sales.pivot_table(
    columns="Month",
    index="Products",
    values="Price",
    aggfunc='sum'
    )
```

```python
shop_sales={
    "products":['laptop','mobile','mouse'],
    "january":[10000,50000,1000],
    "feb":[5500,9999,7776],
    "marh/":[12221,22555,8855]
}
sales=pd.DataFrame(shop_sales)
# melting
sales.melt(id_vars='products', var_name='month', value_name='price')
```
