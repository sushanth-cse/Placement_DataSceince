```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df=pd.read_csv("monthly_sales.csv")
sales_Jan=sales_df[sales_df['Sales Month']=='Jan']
sales_Feb=sales_df[sales_df['Sales Month']=='Feb']
sales_Mar=sales_df[sales_df['Sales Month']=='Mar']
```

```python
#Jan
#Filter Jan Month
sales_Jan=sales_df[sales_df['Sales Month']=='Jan']
sales_Mar=sales_df[sales_df['Sales Month']=='Mar']
sales_df.head(2)
```

No Correlation

```python
#------Month->Jan
#Filter Jan Month Data
sales_Jan=sales_df[sales_df['Sales Month']=='Jan']
#Find Corrrealtion adverstising and sales
sales_Jan['Advertising'].corr(sales_Jan['Sales'])
```

```python
##Positive Correaltion
#Filter Feb Month Data
sales_Feb=sales_df[sales_df['Sales Month']=='Feb']
#Find Corrrealtion adverstising and sales
sales_Feb['Advertising'].corr(sales_Feb['Sales'])
```

Negative Correaltion

```python
##Negative Correaltion
#Filter Mar Month Data
sales_Mar=sales_df[sales_df['Sales Month']=='Mar']
#Find Corrrealtion adverstising and sales
sales_Mar['Advertising'].corr(sales_Mar['Sales'])
```

```python
sns.scatterplot(data=sales_Jan,x='Advertising',y='Sales')
plt.title("Advertising vs Sales for Jan Month")
plt.show()
```

```python
sns.scatterplot(data=sales_Feb,x='Advertising',y='Sales')
plt.title("Advertising vs Sales for Feb Month")
plt.show()
```
