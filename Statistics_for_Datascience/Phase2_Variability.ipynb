```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
titanic_df=pd.read_csv('titanic.csv')
titanic_df.head(5)
```

```python
IT_Department_A = {
    "Employee ID": ["E101", "E102", "E103", "E104", "E105", "E106"],
    "Employee Name": ["Rohan", "Kiran", "Aditya", "Sneha", "Pooja", "Rahul"],
    "Project": [75, 65, 73, 68, 72, 67]
}

IT_Department_B= {
    "Employee ID": ["E201", "E202", "E203", "E204", "E205", "E206"],
    "Employee Name": ["Arjun", "Deepak", "Kavya", "Megha", "Sanjay", "Varun"],
    "Project": [90, 47, 93, 47, 96, 51]
}
deptA_df=pd.DataFrame(IT_Department_A)
deptB_df=pd.DataFrame(IT_Department_B)
```

Range

```python
#Find Range -> IT_Deparment_A
deptA_max=np.max(deptA_df['Project'])
deptA_min=np.min(deptA_df['Project'])
deptA_range=deptA_max-deptA_min
deptA_max,deptA_min,deptA_range
```

```python
#Plot --> IT_Department_A
plt.scatter([i for    i in range(0,len(deptA_df['Project']))],deptA_df['Project'],s=100,marker="o",c="green",label='Dept A')
plt.scatter([i for    i in range(0,len(deptB_df['Project']))],deptB_df['Project'],s=100,marker="*",c="red",label='Dept B')
meanVal=np.mean(deptA_df['Project'])
plt.axhline(y=meanVal,color='yellow',linestyle='--',label='Mean')
plt.legend()
plt.ylim(20,110)
plt.show()
```

```python
#Find MAD Values-> It_deparment_A
x_i=deptA_df['Project']
x_bar_mean=np.mean(deptA_df['Project'])
n=len(deptA_df['Project'])
Mad_deptA=np.sum(np.abs(x_i - x_bar_mean))/n
print("MAD for DEPT B",Mad_deptA)
#Find MAD Values-> It_deparment_A
x_i1=deptB_df['Project']
x_bar_mean1=np.mean(deptB_df['Project'])
n1=len(deptB_df['Project'])
Mad_deptB=np.sum(np.abs(x_i1 - x_bar_mean1))/n1
print("MAD for DEPT B",Mad_deptB)
```
