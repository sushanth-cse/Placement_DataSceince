Percentiles

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

titanic_df=pd.read_csv("titanic.csv")
```

```python
mean_age=titanic_df["Age"].mean()
titanic_df["Age"].fillna(mean_age,inplace=True)
```

```python
#find percentiles at age
print(np.percentile(titanic_df["Age"],0))
print(np.percentile(titanic_df["Age"],25))
print(np.percentile(titanic_df["Age"],50))

print(np.percentile(titanic_df["Age"],75))
print(np.percentile(titanic_df["Age"],100))
```

```python
sns.boxplot(x="Age",data=titanic_df)
```

```python
#find percentiles for Ticket_fare
sns.boxplot(x="Fare",data=titanic_df)
```
