Measures of Cetral Tendency

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns  
import statistics as st   
```

```python
#Data loading
titanic=pd.read_csv("titanic.csv")
titanic.head(10)
```

Mean

```python
meanVal=np.mean(titanic['Fare'])
meanVal
```

```python
meadeanVal=np.median(titanic['Fare'])
meadeanVal
```

```python
sns.histplot(x='Fare',data=titanic,bins=[i for i in range(0,150,10)])
plt.plot([meanVal for i in range(0,351)],
         [i for i in range(0,351)],
         c='red')


plt.ylim(0,400)
plt.show()
```

```python
modeVal=st.mode(titanic['Fare'])
modeVal
```

```python
sns.histplot(x='Fare', data=titanic, bins=[i for i in range(0,150,10)])

plt.plot([meadeanVal for i in range(0,351)],
         [i for i in range(0,351)],
         c='green',label=f'median:{meadeanVal}')
plt.plot([modeVal for i in range(0,351)],
         [i for i in range(0,351)],
         c='orange',label=f'mode:{modeVal}')
plt.legend()
plt.title("Mesuare of central tendency of titanic fare",fontweight='bold',fontsize=15)    


plt.ylim(0,400)
plt.tight_layout()
plt.show()
```
