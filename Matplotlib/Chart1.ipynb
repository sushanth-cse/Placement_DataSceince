Pie Chart

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
weekly_sales= {
    "Brand":["OnePlus","Apple","Oppo","vivo","Samsung"],
    "unit":[23,12,25,28,19]
}
sales_df= pd.DataFrame(weekly_sales)

plt.pie(
    sales_df['unit'],
    labels=sales_df['Brand'],
    autopct='%0.2f%%',
    explode=[0,0,0,0,0.1],
    shadow=True,
    startangle=180
)

plt.title("weekly sales overview", fontsize=12, fontweight='bold')
plt.legend(loc='right',)
plt.show()
```

Histogram

```python
temperature=[28,30,32,31,29,30]
plt.hist(
    temperature,bins=5
    )
plt.title("Day Temperature Overview ",fontsize=15,fontweight="bold",pad=15)
plt.ylim(0,5)
plt.show()
```

Violin Plot

```python
marks=[10,20,30,40,50,50,63,64,68,70,80,90,100]
plt.violinplot(
    marks,showmeans=True
)
plt.grid(axis='y',linestyle="--",alpha=0.4)
plt.show()
```

```python
# Weekly Sales Data (in units)
weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 50]
}
sales =pd.DataFrame(weekly_sales)
plt.stackplot(
    sales['Days'],sales['Week1'], sales['Week2'],sales['Week3'],
    labels=["Week1","Week2","Week3"]

)
plt.grid(axis="both",linestyle="--",alpha=0.9)
plt.legend()
plt.show()
```
