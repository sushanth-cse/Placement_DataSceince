```python
import matplotlib.pyplot as plt
```

Bar Plot

```python
import pandas as pd
```

```python
student_count ={
"Enrolled":[10,20,30,40],
"dept":["cse","civil",'ec','mech']
}
student_df=pd.DataFrame(student_count)

#Create plot
plt.Figure(figsize=(9,5))
bars=plt.bar(student_df["dept"],student_df['Enrolled'],
        color=["#F73F26",'#FF0000', '#0000FF', '#008000', '#FFFF00', '#800080'],edgecolor='green',width=0.6)
#title and  label
plt.title("Dept wise student count",fontsize=13,fontweight='bold',pad=15)
plt.xlabel("Name of Dept",fontsize=12)
plt.ylabel("Department  wise student enrolement",fontsize=12)
#Advance
plt.bar_label(bars,fontsize=12,padding=5,fontweight='bold')
plt.grid(axis='y',linestyle='--',alpha=0.25)
plt.ylim(0,110)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()


plt.show()
```

Line Plot

```python
weekly_sales={
    "Days":['Mon','Tue','wed5','thurs','fri','sat','sun'],
    "Week1_Sales":[32000,43000,50000,60000,70000,750000,900000],
    "Week2_Sales":[35000,43000,55000,65000,75000,750000,950000]

}
sales=pd.DataFrame(weekly_sales)
#----------Week 1 Sales----------
plt.plot(sales['Days'],sales['Week1_Sales'],
         color='red',ls="-",marker='o') #Marker->o,d
#----------Week 2 sales-----------

plt.plot(sales['Days'],sales['Week2_Sales'],
         color='blue',ls="-",marker='o')
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly sales data")
plt.legend()
plt.tight_layout()
plt.ylim(0,100000)

plt.show()

```

Scatter plot

```python
import numpy as np
```

```python
study_hours=(np.random.randint(3,10,50))
marks=np.random.randint(300,500,50)
student_data={
    "Study_hours":study_hours,
    "Marks":marks
}
student_df=pd.DataFrame(student_data)
plt.scatter(student_df["Marks"],student_df["Study_hours"],color="green",alpha=0.5,s=200)
plt.xlabel("Marks of student")
plt.ylabel("Marks of the student")
plt.title("Study vs marks ")
plt.legend()

plt.show()
```
